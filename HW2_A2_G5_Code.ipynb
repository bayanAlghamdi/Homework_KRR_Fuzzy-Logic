### Graphs codes

1. Dust Level – Input 1

```python

import matplotlib.pyplot as plt
import numpy as np
import os

# Create graphs folder
os.makedirs('graphs', exist_ok=True)

# Membership Functions
dust_terms = {
    'Light': (0, 0, 30),
    'Moderate': (0, 30, 100),
    'Heavy': (30, 100, 100)
}

plt.figure(figsize=(10, 6))

# Plot membership functions
for term, (a, b, c) in dust_terms.items():

    # Prepare x and y points
    x = [a, b, c]
    y = [0, 1, 0]

    # Remove first point if a == b
    if a == b:
        x.pop(0)
        y.pop(0)

    # Remove last point if b == c
    if b == c:
        x.pop()
        y.pop()

    # Draw membership function
    plt.plot(x, y, linewidth=2.5, label=term)

# Test values
plt.axvline(20, color='gray', linestyle='--',
            linewidth=1.5, label='Case B (20%)')

plt.axvline(55, color='black', linestyle='--',
            linewidth=1.5, label='Case A (55%)')

# Titles and labels
plt.title('Input 1 - Dust Level Membership Functions',
            fontsize=15, fontweight='bold')

plt.xlabel('Dust Level (%)', fontsize=12)
plt.ylabel('Membership Degree (μ)', fontsize=12)

# Axis limits
plt.xlim(0, 100)
plt.ylim(0, 1.05)

plt.xticks(np.arange(0, 101, 10))
plt.yticks(np.arange(0, 1.1, 0.1))

# Grid and legend
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()

# Save graph
plt.savefig('graphs/Dust_Level.png',
            dpi=300, bbox_inches='tight')

plt.show()
```
